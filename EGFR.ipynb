```python
from chembl_webresource_client.new_client import new_client
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
Target=new_client.target
Target_query=Target.search('EGFR')
Targets=pd.DataFrame.from_dict(Target_query)
Targets
```

```python
target_human=Targets.target_chembl_id[4]
target_human
```

```python
target_human_activity=new_client.activity
res=target_human_activity.filter(target_chembl_id=target_human).filter(standard_type="IC50")
df=pd.DataFrame.from_dict(res)
df
```

```python
df.columns
```

```python
spefic_target=df.filter(['molecule_chembl_id', 'canonical_smiles', 'standard_value', 'standard_units', 'pchembl_value', 'activity_comment', 'assay_type', 'target_chembl_id', 'type'])
spefic_target.to_csv('EGFR.csv',index=False)
```

```python
spefic_target=pd.read_csv('EGFR.csv')
spefic_target
```

```python
remove_nan= spefic_target[spefic_target['standard_units']== 'nM']
remove_nan
```

```python
target_type=spefic_target[spefic_target['type']== 'IC50']
target_type
```

```python
target_type.type.unique()
```

```python
target_type= target_type.dropna(subset=['canonical_smiles', 'standard_value', 'pchembl_value'])
target_type
```

```python
comp_measure = target_type.groupby('molecule_chembl_id').size().reset_index(name='measure_count')
dup_comp= comp_measure[comp_measure['measure_count'] > 1]
dup_comp
```

```python
df_no_duplicates = target_type.drop_duplicates(subset=['molecule_chembl_id'])
display(df_no_duplicates.shape)
display(df_no_duplicates)
```

```python
import numpy as np
target_final= target_type.copy()
target_final['pIC50']= np.log10 (target_type['standard_value']* 1e-9)
display(target_final[['molecule_chembl_id', 'standard_value', 'pIC50']])
```

```python
def bioactivity (IC50_value):
  if IC50_value <= 1000:
    return 'Active'
  elif IC50_value >= 10000:
    return 'Inactive'
  else:
    return 'Intermediate'

target_final['Bioactivity class']= target_final['standard_value'].apply(bioactivity)
target_final
```

```python
target_final['Bioactivity class'].value_counts()
```

```python
target_final_filter= target_final[target_final['Bioactivity class'] != 'Intermediate'].copy()
target_final_filter
```

```python
import rdkit.Chem as Chem
import rdkit.Chem.Descriptors as Descriptors
import pandas as pd
import numpy as np

def calculate_smile(smiles):
  mol = Chem.MolFromSmiles(smiles)
  if mol is None: # Check if mol object was successfully created
    return [np.nan] * 4
  else:
    Mw = Descriptors.MolWt(mol)
    LogP = Descriptors.MolLogP(mol)
    NumHDonors = Descriptors.NumHDonors(mol)
    NumHAcceptors = Descriptors.NumHAcceptors(mol)
    return [Mw, LogP, NumHDonors, NumHAcceptors]

target_final_filter[['Mw', 'LogP', 'NumHDonors', 'NumHAcceptors']]= target_final_filter['canonical_smiles'].apply(lambda x: pd.Series(calculate_smile(x)))
display(target_final_filter)
```
