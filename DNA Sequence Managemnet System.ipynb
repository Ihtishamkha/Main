DNA Sequencing Management System

```python
from typing import Sequence
# DNA Sequence Managemnet System
DNA_database= []
# function to validate DNA seq
def validate_dna(dna_seq):
  dna_seq = dna_seq.upper()
  for base in dna_seq:
    if base not in ['A','T','G','C']:
      return False
  return True

# Add Seq
def add_seq():
  seq_id= input(' Enter Sequence ID:')
  gene_name= input('Enter Gene Name:')
  organism= input('Enter Organism:')
  dna_seq= input('Enter DNA Sequence:').upper()
  if not validate_dna(dna_seq):
    print('Invalid DNA Sequence')
    return
# Checking duplicate Seq Id
  for record in DNA_database:
    if record['Sequence ID']== seq_id:
      print('Duplicate Sequence ID')
      return

  record= {
    'Sequence ID': seq_id,
    'Gene Name': gene_name,
    'Organism': organism,
    'DNA Sequence': dna_seq
  }

  DNA_database.append(record)
  print('Sequence Added Successfully')

#View All Seq
def view_seq():
  if len(DNA_database)==0:
    print('No Sequence Found')
    return
  print('Sequence ID\tGene Name\tOrganism')
  for record in DNA_database:
    print(f"{record['Sequence ID']}\t{record['Gene Name']}\t{record['Organism']}")

 # Search Seq
def search_seq():
    print("\nSearch By")
    print("1. Sequence ID")
    print("2. Gene Name")
    print("3. Organism")
    choice= input("Enter your choice: ")
    found= False
    if choice == '1':
        sid= input('Enter Sequence ID to search: ')
        for record in DNA_database:
            if record['Sequence ID']== sid:
                print(f"Found: {record['Gene Name']} (ID: {record['Sequence ID']})\nOrganism: {record['Organism']}\nDNA Sequence: {record['DNA Sequence']}")
                found = True
                break
        if not found:
            print(f"Sequence with ID '{sid}' not found.")
    elif choice == '2':
        gname = input('Enter Gene Name to search: ')
        for record in DNA_database:
            if record['Gene Name'].lower() == gname.lower():
                print(f"Found: {record['Gene Name']} (ID: {record['Sequence ID']})\nOrganism: {record['Organism']}\nDNA Sequence: {record['DNA Sequence']}")
                found = True
        if not found:
            print(f"Gene with name '{gname}' not found.")
    elif choice == '3':
        org = input('Enter Organism to search: ')
        for record in DNA_database:
            if record['Organism'].lower() == org.lower():
                print(f"Found: {record['Gene Name']} (ID: {record['Sequence ID']})\nOrganism: {record['Organism']}\nDNA Sequence: {record['DNA Sequence']}")
                found = True
        if not found:
            print(f"Organism '{org}' not found.")
    else:
        print('Invalid choice. Please enter 1, 2, or 3.')
```

```python
def main_menu():
    while True:
        print("\n--- DNA Sequence Management System ---")
        print("1. Add New Sequence")
        print("2. View All Sequences")
        print("3. Search Sequence")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            add_seq()
        elif choice == '2':
            view_seq()
        elif choice == '3':
            search_seq()
        elif choice == '4':
            print("Exiting DNA Sequence Management System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")

# Run the main menu
main_menu()
```
